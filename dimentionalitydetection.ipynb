```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Create dataset
data = {
    "Age": [20, 25, 30, 35, 40, 45, 50, 55],
    "Salary": [20000, 25000, 30000, 35000, 40000, 45000, 50000, 55000],
    "Experience": [1, 2, 4, 5, 7, 9, 12, 15],
    "Spending": [80, 75, 70, 65, 60, 55, 50, 45]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Select features
X = df[["Age", "Salary", "Experience", "Spending"]]

# Step 1: Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 2: Apply PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Step 3: Create new DataFrame
pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

print("\nPCA Result:")
print(pca_df)

# Explained variance
print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

# Total information retained
print("\nTotal Variance:")
print(pca.explained_variance_ratio_.sum())

# Visualization
plt.scatter(
    pca_df["PC1"],
    pca_df["PC2"]
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA - Dimensionality Reduction")

plt.show()
```
